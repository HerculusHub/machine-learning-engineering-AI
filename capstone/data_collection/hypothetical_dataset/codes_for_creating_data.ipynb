```python
import pandas as pd
import numpy as np

np.random.seed(42)

# CONFIG
operators = ["Verizon", "AT&T", "T-Mobile"]
regions = ["CA", "TX", "NY", "FL", "IL", "WA", "MA"]
months = pd.date_range("2018-01-01", "2025-12-01", freq="MS")

rows = []

for month in months:
    for region in regions:
        base_population = np.random.randint(5_000_000, 40_000_000)
        
        for op in operators:
            
            # Base characteristics
            price = {
                "Verizon": np.random.normal(75, 3),
                "AT&T": np.random.normal(70, 3),
                "T-Mobile": np.random.normal(65, 3)
            }[op]
            
            promo = np.clip(np.random.normal(6, 2), 0, 10)
            marketing = np.random.uniform(100, 300)
            
            speed = {
                "Verizon": np.random.normal(120, 10),
                "AT&T": np.random.normal(110, 10),
                "T-Mobile": np.random.normal(130, 10)
            }[op]
            
            coverage = {
                "Verizon": np.random.normal(95, 2),
                "AT&T": np.random.normal(93, 2),
                "T-Mobile": np.random.normal(90, 3)
            }[op]
            
            # Seasonality
            holiday = 1 if month.month in [11, 12] else 0
            
            # Demand model
            demand = (
                50000
                - 300 * price
                + 2000 * np.log(marketing)
                + 5000 * promo
                + 800 * speed
                + 1000 * holiday
            )
            
            net_adds = int(np.random.normal(demand, 10000))
            
            churn = np.clip(
                2.5 - 0.01 * coverage + 0.02 * price - 0.1 * promo,
                0.5, 5
            )
            
            subscribers = int(
                base_population * np.random.uniform(0.2, 0.4)
            )
            
            satisfaction = np.clip(
                5 + 0.02 * speed + 0.03 * coverage - 0.01 * price,
                0, 10
            )
            
            rows.append([
                month, region, op, subscribers, net_adds, churn,
                price, promo, marketing, speed, coverage, satisfaction
            ])

df = pd.DataFrame(rows, columns=[
    "month", "region", "operator", "subscribers", "net_adds",
    "churn_rate", "avg_price", "promo_intensity", "marketing_spend",
    "network_speed", "coverage_score", "satisfaction"
])

df.head()
```
