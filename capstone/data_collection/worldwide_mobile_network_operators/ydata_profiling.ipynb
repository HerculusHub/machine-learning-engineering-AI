```python
# Load data
df = pd.read_csv(
    r"...\telecom_customer_churn_case_study\DataDictionary-TelecomChurnCaseStudy.csv"
)
```

```python
print(df.head())
```

```python
# Create profile
profile = ProfileReport(df, title="report")

# Get profile description
profile_data = profile.get_description()

# Create markdown text
md = "# Data Profile Report\n\n"

md += "## Overview\n"

md += f"- Number of rows: {profile_data.table['n']}\n"
md += f"- Number of columns: {profile_data.table['n_var']}\n"
md += f"- Missing cells: {profile_data.table['n_cells_missing']}\n\n"

md += "## Columns\n\n"

for col, details in profile_data.variables.items():

    md += f"### {col}\n"

    if "type" in details:
        md += f"- Type: {details['type']}\n"

    if "mean" in details:
        md += f"- Mean: {details['mean']}\n"

    if "min" in details:
        md += f"- Min: {details['min']}\n"

    if "max" in details:
        md += f"- Max: {details['max']}\n"

    md += "\n"

# Save markdown file
with open("report.md", "w", encoding="utf-8") as f:
    f.write(md)

print("Markdown report created!")
```
